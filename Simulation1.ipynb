```python
import pandas as pd
import numpy as np
```

```python
stage = pd.read_json('full_stage_data.json')
```

```python
stage.drop(['indices','feature_type'],axis=1, inplace=True)
stage
```

```python
def weighted_mean(values, weights):
    return (values * weights).sum() / weights.sum()
```

```python
global_value = stage[['Stamina',
       'Sprint', 'Climbing', 'Flat', 'Technique', 'Downhill', 'Hills',
       'Aggressiveness', 'Teamwork']].agg(
            Stamina=('Stamina', lambda x: weighted_mean(x, stage.loc[x.index, 'difficulty'])),
            Sprint=('Sprint', lambda x: weighted_mean(x, stage.loc[x.index, 'difficulty'])),
            Climbing=('Climbing', lambda x: weighted_mean(x, stage.loc[x.index, 'difficulty'])),
            Flat=('Flat', lambda x: weighted_mean(x, stage.loc[x.index, 'difficulty'])),
            Technique=('Technique', lambda x: weighted_mean(x, stage.loc[x.index, 'difficulty'])),
            Downhill=('Downhill', lambda x: weighted_mean(x, stage.loc[x.index, 'difficulty'])),
            Hills=('Hills', lambda x: weighted_mean(x, stage.loc[x.index, 'difficulty'])),
            Aggressiveness=('Aggressiveness', lambda x: weighted_mean(x, stage.loc[x.index, 'difficulty'])),
            Teamwork=('Teamwork', lambda x: weighted_mean(x, stage.loc[x.index, 'difficulty']))
       )[['Stamina',
       'Sprint', 'Climbing', 'Flat', 'Technique', 'Downhill', 'Hills',
       'Aggressiveness', 'Teamwork']].max()
global_value
```

```python
cyclists = pd.read_json('ciclistas.json')
cyclists
```

```python
mappings = {
    'resistencia':'Stamina',
    'sprint':'Sprint',
    'montana':'Climbing',
    'llano':'Flat',
    'tecnica':'Technique',
    'descenso':'Downhill',
    'colinas':'Hills',
    'agresividad':'Aggressiveness',
    'trabajoequipo':'Teamwork'
}

cyclists.rename(columns=mappings, inplace=True)
cyclists
```

```python
def compute_global_value(df, global_abilities):
    # Calculate cross product (element-wise product for matching columns)
    numeric_cols = global_abilities.index  # Columns to use from the series
    df['global_value'] = df[numeric_cols].dot(global_abilities)

    df_sorted = df.sort_values(by='global_value', ascending=False)
    
    return df_sorted
```

```python
compute_global_value(cyclists, global_value)
```
